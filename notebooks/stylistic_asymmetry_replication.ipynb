```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/stylistic_asymmetry_complete_dataset.csv')
df.head()
```

### 1. Stylometric Summary Statistics

```python
metrics = ['MATTR','Burrows_Delta','MLS','HLR','Mean_Word_Len','Lexical_Density']
df.groupby(['Tradition','Type'])[metrics].agg(['mean','std']).round(3)
```

### 2. Stylistic Divergence Gradient

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
gpt_df = df[df['Type'] == 'GPT-4o']
sns.boxplot(ax=ax[0], data=gpt_df, x='Tradition', y='Burrows_Delta', palette=['#64748B','#F59E0B','#EF4444'], width=0.4)
sns.stripplot(ax=ax[0], data=gpt_df, x='Tradition', y='Burrows_Delta', color='black', alpha=0.7, jitter=0.1)
ax[0].set_title("Burrows' Delta across Traditions (GPT-4o)", fontweight='bold')
sns.barplot(ax=ax[1], data=df, x='Tradition', y='MLS', hue='Type', palette=['#1F2937','#2563EB'], errorbar='sd', capsize=0.1)
ax[1].set_title('Mean Length of Sentence (MLS)', fontweight='bold')
plt.tight_layout(); plt.show()
```
